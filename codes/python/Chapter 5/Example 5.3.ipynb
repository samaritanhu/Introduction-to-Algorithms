input: adj, n, k
output: U (the set of vertices of the maximal induced subgraph in which every vertex has degree >= k)

```python
import queue

def MaximalInducedSubgraph(adj, n, k):
    '''
    adj: adjacency list, adj[i] is an iterable of the neighbors of vertex i (0-indexed)
    n: number of vertices
    k: every remaining vertex must have degree >= k in the induced subgraph

    Returns the set U of vertices (0-indexed) of the maximal induced subgraph
    H = (U, F) such that every vertex in H has degree >= k. U is empty if no
    such non-empty subgraph exists.
    '''
    degree = [len(adj[i]) for i in range(n)]
    removed = [False] * n
    Queue = queue.Queue()
    for i in range(n):
        if degree[i] < k:
            removed[i] = True
            Queue.put(i)
    while not Queue.empty():
        u = Queue.get()
        for v in adj[u]:
            if not removed[v]:
                degree[v] -= 1
                if degree[v] < k:
                    removed[v] = True
                    Queue.put(v)
    return {i for i in range(n) if not removed[i]}
```

```python
# 8 scientists 0..7. Edges mean two scientists are willing to talk to each other.
# 0-1-2 form a triangle, 2-3 bridges to another triangle 3-4-5, and 5-6-7 is a pendant path.
adj = {
    0: [1, 2],
    1: [0, 2],
    2: [0, 1, 3],
    3: [2, 4, 5],
    4: [3, 5],
    5: [3, 4, 6],
    6: [5, 7],
    7: [6],
}
n = 8
k = 2
print(MaximalInducedSubgraph(adj, n, k))
```
