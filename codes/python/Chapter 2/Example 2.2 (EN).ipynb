Theorem 2.2, 2.3, 2.4 - three warm-up induction proofs from section 2.2.

input: n (2.2, 2.3), or n and x (2.4)

output: the closed form each theorem claims, checked against a direct computation (not just trusted)

```python
def sum_first_n(n):
    '''
    Theorem 2.2: 1 + 2 + ... + n = n(n+1)/2
    '''
    closed_form = n * (n + 1) // 2
    direct_sum = sum(range(1, n + 1))
    assert closed_form == direct_sum
    return closed_form


def series_T(n):
    '''
    Theorem 2.3: 8 + 13 + 18 + 23 + ... + (3 + 5n) = 2.5n^2 + 5.5n
    '''
    closed_form = 2.5 * n ** 2 + 5.5 * n
    direct_sum = sum(3 + 5 * i for i in range(1, n + 1))
    assert closed_form == direct_sum
    return closed_form


def bernoulli(n, x):
    '''
    Theorem 2.4 (Bernoulli's inequality): for natural n and 1 + x > 0,
    (1 + x)^n >= 1 + nx
    '''
    assert 1 + x > 0
    return (1 + x) ** n >= 1 + n * x
```

```python
for n in range(1, 11):
    print(f"n={n:2d}  S(n)={sum_first_n(n):3d}  T(n)={series_T(n)}")

print()
for x in [-0.5, 0, 0.3, 2, 5]:
    for n in range(1, 6):
        assert bernoulli(n, x)
    print(f"x={x}: (1+x)^n >= 1+nx holds for n=1..5")
```
