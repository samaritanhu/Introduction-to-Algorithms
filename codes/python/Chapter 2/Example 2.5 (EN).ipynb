A More Complex Summation Problem - the odd-number triangle

```
1               = 1
3 + 5           = 8
7 + 9 + 11      = 27
13+15+17+19     = 64
21+23+25+27+29  = 125
```

Row i holds i consecutive odd numbers, continuing straight on from row i-1.
Row sums look like cubes; claim is row i sums to i^3.

input: n (number of rows)

output: two independent builds of the same triangle cross-checked against each other
and against i^3 - one generates rows sequentially handing out odd numbers in order
(how the picture is actually built), the other jumps straight to row i via the closed
form for its first entry that the book's induction proof derives: i^2 - i + 1

````python
def generate_triangle(n):
    '''Sequential build: hand out consecutive odd numbers, i per row.'''
    rows = []
    next_odd = 1
    for i in range(1, n + 1):
        row = [next_odd + 2 * k for k in range(i)]
        rows.append(row)
        next_odd += 2 * i
    return rows


def triangle_row(i):
    '''
    Direct jump to row i via its first entry, i^2 - i + 1 - the value the
    proof's nested induction lemma (last entry of row i is i^2 + i - 1) implies.
    '''
    start = i * i - i + 1
    return [start + 2 * k for k in range(i)]


def verify(n):
    rows = generate_triangle(n)
    for i in range(1, n + 1):
        row = rows[i - 1]
        assert row == triangle_row(i)
        assert sum(row) == i ** 3
    return rows
````

````python
rows = verify(5)
for i, row in enumerate(rows, start=1):
    print(f"row {i}: {' + '.join(map(str, row))} = {sum(row)} (i^3 = {i**3})")
````
