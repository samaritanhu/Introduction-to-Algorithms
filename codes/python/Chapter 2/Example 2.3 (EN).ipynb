Theorem 2.5 - counting regions n lines in general position cut the plane into.

input: n (number of lines, no two parallel, no three meeting at one point)

output: region count, computed two ways and cross-checked - the recurrence R(n) = R(n-1) + n
(add one line, it crosses every existing line once, splitting n existing regions in two) against
the closed form R(n) = n(n+1)/2 + 1 the proof derives from it

```python
def regions_by_recurrence(n):
    '''
    R(1) = 2, R(i) = R(i-1) + i for i > 1.
    Direct translation of the induction step: the i-th line meets the i-1
    earlier lines in i-1 points, splitting it into i pieces, each piece
    cutting one existing region into two, so i new regions appear.
    '''
    regions = 2
    for i in range(2, n + 1):
        regions += i
    return regions if n >= 1 else 1


def regions_closed_form(n):
    '''
    Theorem 2.5: n(n+1)/2 + 1
    '''
    return n * (n + 1) // 2 + 1


def regions(n):
    a, b = regions_by_recurrence(n), regions_closed_form(n)
    assert a == b
    return a
```

```python
for n in range(1, 11):
    print(f"n={n:2d} lines -> {regions(n):3d} regions")
```
