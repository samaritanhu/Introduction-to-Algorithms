定理 2.2、2.3、2.4 —— 2.2 节三个简单的归纳法例子。

input: n（定理2.2、2.3），或 n 和 x（定理2.4）

output: 每个定理给出的闭式解，与直接计算结果做校验（不直接相信闭式解）

```python
def sum_first_n(n):
    '''
    定理 2.2: 1 + 2 + ... + n = n(n+1)/2
    '''
    closed_form = n * (n + 1) // 2
    direct_sum = sum(range(1, n + 1))
    assert closed_form == direct_sum
    return closed_form


def series_T(n):
    '''
    定理 2.3: 8 + 13 + 18 + 23 + ... + (3 + 5n) = 2.5n^2 + 5.5n
    '''
    closed_form = 2.5 * n ** 2 + 5.5 * n
    direct_sum = sum(3 + 5 * i for i in range(1, n + 1))
    assert closed_form == direct_sum
    return closed_form


def bernoulli(n, x):
    '''
    定理 2.4（Bernoulli不等式）: 若 n 为自然数且 1 + x > 0，
    则 (1 + x)^n >= 1 + nx
    '''
    assert 1 + x > 0
    return (1 + x) ** n >= 1 + n * x
```

```python
for n in range(1, 11):
    print(f"n={n:2d}  S(n)={sum_first_n(n):3d}  T(n)={series_T(n)}")

print()
for x in [-0.5, 0, 0.3, 2, 5]:
    for n in range(1, 6):
        assert bernoulli(n, x)
    print(f"x={x}: (1+x)^n >= 1+nx 在 n=1..5 时均成立")
```
