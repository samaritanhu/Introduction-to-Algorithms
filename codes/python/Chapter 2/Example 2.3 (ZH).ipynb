定理 2.5 —— n 条居一般位置的直线把平面分割成的区域数。

input: n（直线数目，任意两条不平行、任意三条不共点）

output: 区域数，用两种方式计算并互相校验——递推式 R(n) = R(n-1) + n
（新增一条直线，它与已有每条直线各交于一点，把 n 个已有区域各一分为二）
与证明由此推出的闭式解 R(n) = n(n+1)/2 + 1

```python
def regions_by_recurrence(n):
    '''
    R(1) = 2，R(i) = R(i-1) + i（i > 1）。
    直接对应归纳步骤：第 i 条直线与前 i-1 条直线相交于 i-1 个点，
    被分成 i 段，每段把一个已有区域切成两个，因此新增 i 个区域。
    '''
    regions = 2
    for i in range(2, n + 1):
        regions += i
    return regions if n >= 1 else 1


def regions_closed_form(n):
    '''
    定理 2.5: n(n+1)/2 + 1
    '''
    return n * (n + 1) // 2 + 1


def regions(n):
    a, b = regions_by_recurrence(n), regions_closed_form(n)
    assert a == b
    return a
```

```python
for n in range(1, 11):
    print(f"n={n:2d} 条直线 -> {regions(n):3d} 个区域")
```
