复杂一些的加法题 —— 奇数三角形

```
1               = 1
3 + 5           = 8
7 + 9 + 11      = 27
13+15+17+19     = 64
21+23+25+27+29  = 125
```

第 i 行有 i 个连续奇数，紧接第 i-1 行往下排。每行的和看起来都是立方数，
猜测：第 i 行之和为 i^3。

input: n（行数）

output: 用两种独立方式构造同一个三角形并互相校验、再对照 i^3 ——
一种按顺序逐行发放奇数（图本身的生成方式），另一种直接跳到第 i 行，
用书中归纳证明推出的首项闭式解 i^2 - i + 1 定位该行

````python
def generate_triangle(n):
    '''顺序构造：按行发放连续奇数，第 i 行发 i 个。'''
    rows = []
    next_odd = 1
    for i in range(1, n + 1):
        row = [next_odd + 2 * k for k in range(i)]
        rows.append(row)
        next_odd += 2 * i
    return rows


def triangle_row(i):
    '''
    直接用首项闭式解定位第 i 行：i^2 - i + 1。
    该值由证明中的嵌套归纳引理（第 i 行末项为 i^2 + i - 1）推出。
    '''
    start = i * i - i + 1
    return [start + 2 * k for k in range(i)]


def verify(n):
    rows = generate_triangle(n)
    for i in range(1, n + 1):
        row = rows[i - 1]
        assert row == triangle_row(i)
        assert sum(row) == i ** 3
    return rows
````

````python
rows = verify(5)
for i, row in enumerate(rows, start=1):
    print(f"第{i}行: {' + '.join(map(str, row))} = {sum(row)} (i^3 = {i**3})")
````
